```python
import os
import zipfile
from getpass import getpass
import kagglehub

DATA_DIR = '/content/data'
DRIVE_DIR = '/content/drive/MyDrive/dl-qpi/data'

# RBCs Holograms
ZIP_NAME_RBC = 'RBCs Holograms.zip'
ZIP_PATH_RBC = os.path.join(DRIVE_DIR, ZIP_NAME_RBC)
DOWNLOAD_URL = 'https://osf.io/download/cp2kn/?view_only=e8d752fda65e456f896b72a91aa8db51'
EXTRACT_DIR_RBC = '/content/data/rbc_holograms'

# P.Vivax Malaria
ZIP_NAME_MALARIA = 'P_Vivax_Malaria.zip'
ZIP_PATH_MALARIA = os.path.join(DRIVE_DIR, ZIP_NAME_MALARIA)
KAGGLE_DATASET = 'orvile/p-vivax-malaria-infected-human-blood-smears'
EXTRACT_DIR_MALARIA = '/content/data/pvivax_malaria'
```

## Descargas a drive (cache estable)

```python
# RBCs Holograms
os.makedirs(DRIVE_DIR, exist_ok=True)

if os.path.exists(ZIP_PATH_RBC):
    size_gb = os.path.getsize(ZIP_PATH_RBC) / (1024 ** 3)
    print(f'Ya existe: {ZIP_PATH_RBC} ({size_gb:.2f} GB). No se descarga de nuevo.')
else:
    print(f'Descargando a {ZIP_PATH_RBC} ...')
    !wget -O "{ZIP_PATH_RBC}" "{DOWNLOAD_URL}"
```

```python
# P.Vivax Malaria
os.makedirs(DRIVE_DIR, exist_ok=True)

if os.path.exists(ZIP_PATH_MALARIA):
    size_gb = os.path.getsize(ZIP_PATH_MALARIA) / (1024 ** 3)
    print(f'Ya existe: {ZIP_PATH_MALARIA} ({size_gb:.2f} GB). No se descarga de nuevo.')
else:
    kaggle_token = getpass("Kaggle API token (KGAT_...): ")

    kaggle_dir = os.path.expanduser("~/.kaggle")
    os.makedirs(kaggle_dir, exist_ok=True)
    token_path = os.path.join(kaggle_dir, "access_token")
    with open(token_path, "w") as f:
        f.write(kaggle_token.strip())
    os.chmod(token_path, 0o600)

    # El CLI de kaggle deja el .zip sin descomprimir (kagglehub lo extrae siempre)
    print(f'Descargando a {ZIP_PATH_MALARIA} ...')
    !kaggle datasets download -d "{KAGGLE_DATASET}" -p "{DRIVE_DIR}"
    os.rename(os.path.join(DRIVE_DIR, KAGGLE_DATASET.split('/')[1] + '.zip'), ZIP_PATH_MALARIA)
```

## Extracción a runtime de Colab

```python
# RBCs Holograms.zip

os.makedirs(EXTRACT_DIR_RBC, exist_ok=True)

if os.listdir(EXTRACT_DIR_RBC):
    print(f'{EXTRACT_DIR_RBC} ya tiene contenido, se omite la descompresión.')
else:
    print(f'Descomprimiendo {ZIP_PATH_RBC} en {EXTRACT_DIR_RBC} ...')
    with zipfile.ZipFile(ZIP_PATH_RBC, 'r') as zf:
        zf.extractall(EXTRACT_DIR_RBC)
    print('Listo.')
```

```python
# P.Vivax Malaria.zip

os.makedirs(EXTRACT_DIR_MALARIA, exist_ok=True)

if os.listdir(EXTRACT_DIR_MALARIA):
    print(f'{EXTRACT_DIR_MALARIA} ya tiene contenido, se omite la descompresión.')
else:
    print(f'Descomprimiendo {ZIP_PATH_MALARIA} en {EXTRACT_DIR_MALARIA} ...')
    with zipfile.ZipFile(ZIP_PATH_MALARIA, 'r') as zf:
        zf.extractall(EXTRACT_DIR_MALARIA)
    print('Listo.')
```
